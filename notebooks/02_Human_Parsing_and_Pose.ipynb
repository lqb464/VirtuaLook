# 02 — Human Parsing and Pose Estimation

**Goal**: Use MediaPipe to extract human pose keypoints and generate an agnostic mask (removing the current clothes so the model can draw the new ones).

```python
import cv2
import mediapipe as mp
import numpy as np
import matplotlib.pyplot as plt

mp_pose = mp.solutions.pose
pose = mp_pose.Pose(static_image_mode=True, model_complexity=2, enable_segmentation=True)
```

```python
# Dummy image processing
img = np.zeros((512, 512, 3), dtype=np.uint8)
results = pose.process(img)

print("Pose landmarks detected:", results.pose_landmarks is not None)
print("Segmentation mask generated:", results.segmentation_mask is not None)
```

The segmentation mask is critical. We use it to create an `agnostic mask` that tells the Diffusion model: "Do not touch the face, hands, or background. Only regenerate the torso/shirt area."
