# 01 — VITON-HD Data Preparation

**Goal**: Explore and preprocess the VITON-HD dataset. 
The dataset contains pairs of images: a person wearing an outfit, and a flat-lay image of the target garment.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

In a real environment, you would download the dataset, map the person image to the garment image, and create a DataLoader.

```python
def create_dummy_pair():
    person = np.ones((512, 384, 3), dtype=np.uint8) * 200
    garment = np.ones((512, 384, 3), dtype=np.uint8) * 100
    return person, garment

person, garment = create_dummy_pair()

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(person)
ax[0].set_title("Person Image")
ax[1].imshow(garment)
ax[1].set_title("In-shop Garment Image")
plt.show()
```
